```python
# Đếm số lượng tháng của mỗi customer
customer_coverage = (
    df.groupby("consumer_id")["analysis_month"]
    .apply(lambda x: x.dt.to_period("M").nunique())
)

print("=== Phân phối số tháng dữ liệu ===")
print(customer_coverage.value_counts().sort_index())

total_customers = customer_coverage.shape[0]
full_coverage_customers = (customer_coverage == 12).sum()
full_coverage_pct = full_coverage_customers / total_customers * 100 # tính tỷ lệ phần trăm

print(f"\nTổng số customer: {total_customers:,}")
print(f"Customer có đủ 12 tháng: {full_coverage_customers:,} ({full_coverage_pct:.2f}%)")
print(f"Customer thiếu tháng: {total_customers - full_coverage_customers:,} "
      f"({100 - full_coverage_pct:.2f}%)")
```

### Kiểm tra giá trị ngoại lai

```python
# các biến liên tục
numeric_cols = [
    "monthly_income_vnd",
    "credit_limit_vnd",
    "opening_balance_vnd",
    "ending_balance_vnd",
    "total_spend_vnd",
    "essential_spend_vnd",
    "discretionary_spend_vnd",
    "online_spend_vnd",
    "transaction_count",
    "active_transaction_days",
    "average_transaction_value_vnd",
    "category_diversity",
    "spend_to_income_ratio",
    "credit_utilization_ratio",
    "transaction_recency_days",
    "spending_volatility"
]
```

```python
# xác định các giá trị ngoại lai
def iqr_outlier_summary(df: pd.DataFrame, columns: list[str]) -> pd.DataFrame:
    results = []
    for col in columns:
        x = df[col].dropna()

        q1 = x.quantile(0.25)
        q3 = x.quantile(0.75)
        iqr = q3 - q1

        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr

        n_outliers = ((x < lower) | (x > upper)).sum()

        results.append({
            "variable": col,
            "Q1": q1,
            "Q3": q3,
            "lower_bound": lower,
            "upper_bound": upper,
            "outlier_count": n_outliers,
            "outlier_pct": n_outliers / len(x) * 100 # tỷ lệ ngoại lai so với tổng
        })

    return pd.DataFrame(results)


outlier_report = iqr_outlier_summary(
    df_balanced,
    numeric_cols
)

display(outlier_report)
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

sorted_report = outlier_report.sort_values("outlier_pct", ascending=True)

ax.barh(sorted_report["variable"], sorted_report["outlier_pct"], color="#2a78d6")
ax.set_xlabel("Outlier (%)")
ax.set_title("Proportion of outliers by variable (IQR method)")

# Thêm nhãn số liệu trên mỗi bar
for i, v in enumerate(sorted_report["outlier_pct"]):
    ax.text(v + 0.1, i, f"{v:.2f}%", va="center", fontsize=9)

plt.tight_layout()
plt.show()
```

```python
n_cols = len(numeric_cols)
n_rows = int(np.ceil(n_cols / 3))
fig, axes = plt.subplots(n_rows, 3, figsize=(15, 4 * n_rows))
axes = axes.flatten()

for i, col in enumerate(numeric_cols):
    x = df[col].dropna()
    row = outlier_report[outlier_report["variable"] == col].iloc[0]

    sns.histplot(x, ax=axes[i], bins=40, color="#2a78d6", kde=True)
    axes[i].axvline(row["lower_bound"], color="#e34948", linestyle="--", linewidth=1, label="lower bound")
    axes[i].axvline(row["upper_bound"], color="#e34948", linestyle="--", linewidth=1, label="upper bound")
    axes[i].set_title(f"{col} (outliers: {row['outlier_pct']:.1f}%)", fontsize=10)
    axes[i].legend(fontsize=7)

for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
plt.show()
```

#### Nhận xét 4:
Ta thấy phần lớn dữ liệu đều có ngoại lai, tuy nhiên, *category_diversity* có tỷ lệ cao đột biến. Phần lớn khách hàng chi tiêu đa dạng các nhóm, tuy nhiên, một số khách hàng chỉ chi tiêu có số ít nhóm.

## Phần 2: Phân tích định lượng

### Q1: Phân tích tháng có chi tiêu cao nhất và tháng có chi tiêu thấp nhất để phân tích hành vi khách hàng trong thời gian cao điểm

```python
# tổng chi tiêu và transaction theo tháng
monthly = (df_balanced.groupby("analysis_month").agg(
          total_spend=("total_spend_vnd", "sum"),
          transaction_count=("transaction_count", "sum")
      )
)
# trung bình chi tiêu cho mỗi giao dịch
monthly["average_ticket"] = (monthly["total_spend"]/ monthly["transaction_count"])
# thời gian có tổng chi tiêu cao nhất
peak = monthly["total_spend"].idxmax()
# thời gian có tổng chi tiêu thấp nhất
lowest = monthly["total_spend"].idxmin()

annual_spend = monthly["total_spend"].sum()
# tỷ lệ phần trăm của tháng so với cả năm
monthly["annual_spend_share"] = (
    monthly["total_spend"] / annual_spend * 100
)
```

```python
display(monthly)
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import pandas as pd
import numpy as np
from datetime import timedelta
# Đảm bảo index là datetime
monthly = monthly.copy()
monthly.index = pd.to_datetime(monthly.index)

peak_month = monthly["total_spend"].idxmax()
lowest_month = monthly["total_spend"].idxmin()

fig, ax = plt.subplots(figsize=(12, 6))

colors = [
    "#e34948" if m == peak_month else "#117f78" if m == lowest_month else "#2a78d6"
    for m in monthly.index
]

bars = ax.bar(monthly.index, monthly["total_spend"] / 1e9, color=colors, width=20)

# Format trục X theo tháng
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b"))
ax.xaxis.set_major_locator(mdates.MonthLocator())
plt.xticks(rotation=45)

ax.set_ylabel("Total expenditure (VND billion)", fontweight="bold")
ax.set_title("Total monthly expenditure in 2025", fontsize=18, fontweight="bold", loc="left")
ax.set_xlim(monthly.index.min() - timedelta(days=20),
            monthly.index.max() + timedelta(days=20))
max_val = (monthly["total_spend"] / 1e9).max()
ax.set_ylim(0, max_val * 1.15)
# Annotate peak & lowest
ax.annotate(
    f"Highest\n({monthly.loc[peak_month,'annual_spend_share']:.1f}% year)",
    xy=(peak_month, monthly.loc[peak_month, "total_spend"] / 1e9),
    xytext=(0, 15), textcoords="offset points",
    ha="center", fontsize=9, fontweight="bold", color="#e34948",
)
ax.annotate(
    f"Lowest\n({monthly.loc[lowest_month,'annual_spend_share']:.1f}% year)",
    xy=(lowest_month, monthly.loc[lowest_month, "total_spend"] / 1e9),
    xytext=(0, 15), textcoords="offset points",
    ha="center", fontsize=9, fontweight="bold", color="#117f78",
)

# Đường trung bình tham chiếu
avg_spend = monthly["total_spend"].mean() / 1e9
ax.axhline(avg_spend, color="#b4b5b6", linestyle="--", linewidth=1, alpha=0.7)
ax.text(monthly.index[0], avg_spend * 1.02, f"Average: {avg_spend:,.0f} billion", fontsize=8, color="#b4b5b6", fontweight="bold")

plt.tight_layout(rect=[0, 0, 1, 1])
plt.show()
```

```python
# tháng có chi tiêu cao nhất tính theo VNĐ và %
peak_total_spend = monthly.loc[peak, "total_spend"]
peak_share = monthly.loc[peak, "annual_spend_share"]
# tháng có chi tiêu thấp nhất tính theo VNĐ và %
lowest_total_spend = monthly.loc[lowest, "total_spend"]
lowest_share = monthly.loc[lowest, "annual_spend_share"]
# số lượng transaction của tháng có chi tiêu cao nhất và chi tiêu thấp nhất (tính theo số lượng và tỷ lệ % so với cả năm)
peak_txn_count = monthly.loc[peak, "transaction_count"]
lowest_txn_count = monthly.loc[lowest, "transaction_count"]
txn_count_diff = peak_txn_count - lowest_txn_count
txn_count_diff_pct = (txn_count_diff / lowest_txn_count) * 100
# trung bình chi tiêu của giao dịch của tháng có chi tiêu cao nhất và chi tiêu thấp nhất
peak_avg_ticket = monthly.loc[peak, "average_ticket"]
lowest_avg_ticket = monthly.loc[lowest, "average_ticket"]
# tỷ lệ chênh lệch giữa trung bình chi tiêu của giao dịch giữa tháng có chi tiêu cao so với tháng có chi tiêu thấp
avg_ticket_diff = peak_avg_ticket - lowest_avg_ticket
avg_ticket_diff_pct = (avg_ticket_diff / lowest_avg_ticket) * 100
```

```python
print(f"=== Highest month: {peak} ===")
print(f"Total spend: {peak_total_spend:,.0f} VND")
print(f"Share of annual spend: {peak_share:.2f}%")
print(f"Transaction count: {peak_txn_count:,.0f}")
print(f"Average ticket: {peak_avg_ticket:,.0f} VND")
```

```python
print(f"\n=== Lowest month: {lowest} ===")
print(f"Total spend: {lowest_total_spend:,.0f} VND")
print(f"Share of annual spend: {lowest_share:.2f}%")
print(f"Transaction count: {lowest_txn_count:,.0f}")
print(f"Average ticket: {lowest_avg_ticket:,.0f} VND")
```

```python
print(f"\n=== Peak vs Lowest ===")
print(f"Transaction count diff: {txn_count_diff:,.0f} ({txn_count_diff_pct:+.1f}%)")
print(f"Average ticket diff: {avg_ticket_diff:,.0f} VND ({avg_ticket_diff_pct:+.1f}%)")
```

```python
import matplotlib.pyplot as plt
import numpy as np

COLOR_HIGHEST = "#e34948"
COLOR_LOWEST = "#117f78"

compare_months = monthly.loc[[lowest_month, peak_month]]

fig, ax1 = plt.subplots(figsize=(8, 5))

x = np.arange(2)
width = 0.35
labels = [
    lowest_month.strftime("%b-%Y") + "\n(Lowest)",
    peak_month.strftime("%b-%Y") + "\n(Highest)"
]


bars1 = ax1.bar(x - width/2, compare_months["transaction_count"], width,
                 color=COLOR_HIGHEST, label="Transaction count")
ax1.set_ylabel("Transaction count", color=COLOR_HIGHEST, fontweight="bold")
ax1.tick_params(axis="y", labelcolor=COLOR_HIGHEST)

ax2 = ax1.twinx()
bars2 = ax2.bar(x + width/2, compare_months["average_ticket"] / 1e6, width,
                 color=COLOR_LOWEST, label="Average ticket (VND million)")
ax2.set_ylabel("Average ticket (VND million)", color=COLOR_LOWEST, fontweight="bold")
ax2.tick_params(axis="y", labelcolor=COLOR_LOWEST)


ax1.set_xticks(x)
ax1.set_xticklabels(labels, fontweight="bold")
ax1.set_title("Driver comparison: Transaction count vs. Average ticket",
              fontsize=13, fontweight="bold", loc="left")

max_txn = compare_months["transaction_count"].max()
ax1.set_ylim(0, max_txn * 1.15)

max_ticket = (compare_months["average_ticket"] / 1e6).max()
ax2.set_ylim(0, max_ticket * 1.15)

# label
for b in bars1:
    ax1.text(b.get_x() + b.get_width()/2, b.get_height() + (max_txn * 0.02),
             f"{b.get_height():,.0f}", ha="center", fontsize=9, fontweight="bold", color=COLOR_HIGHEST)

for b in bars2:
    ax2.text(b.get_x() + b.get_width()/2, b.get_height() + (max_ticket * 0.02),
             f"{b.get_height():,.2f}M", ha="center", fontsize=9, fontweight="bold", color=COLOR_LOWEST)

# tính chênh lệch
txn_diff_pct = (compare_months.loc[peak_month, "transaction_count"] /
                compare_months.loc[lowest_month, "transaction_count"] - 1) * 100
ticket_diff_pct = (compare_months.loc[peak_month, "average_ticket"] /
                   compare_months.loc[lowest_month, "average_ticket"] - 1) * 100

fig.text(0.5, 0.02,
         f"Transaction count changed {txn_diff_pct:+.1f}%  |  Average ticket changed {ticket_diff_pct:+.1f}%",
         ha="center", fontsize=10, fontweight="bold")

plt.tight_layout(rect=[0, 0.05, 1, 0.93])
plt.show()
```

```python
# chia nhóm độ tuổi của vòng đời con người
'''
<25: mới đi làm/sinh viên, thu nhập thấp-không ổn định, ít tài sản, dùng thẻ tín dụng/vay tiêu dùng lần đầu.
25-34: giai đoạn ổn định công việc, bắt đầu tích lũy, nhiều người lập gia đình — chi tiêu tăng mạnh (cưới hỏi, mua sắm nhà cửa ban đầu).
35-44: đỉnh cao chi tiêu gia đình (nuôi con nhỏ, mua nhà/xe, vay thế chấp) — nhóm quan trọng nhất với công ty tài chính.
45-54: thu nhập ổn định hơn nhưng áp lực tài chính kép (con cái tuổi đi học, cha mẹ già).
55-64: chuẩn bị nghỉ hưu, xu hướng giảm vay/tăng tiết kiệm.
65+: nghỉ hưu, thu nhập giảm, mô hình chi tiêu khác biệt hẳn (y tế chiếm tỷ trọng lớn)
'''
bins = [13, 24, 34, 44, 54, 64, np.inf]
labels = ["<25", "25-34", "35-44", "45-54", "55-64", "65+"]

df_balanced["age_cohort"] = pd.cut(df["age"], bins=bins, labels=labels)

age_summary = (
    df_balanced.groupby("age_cohort", observed=True)
      .agg(
          avg_transaction_count=("transaction_count", "mean"), # số lượng giao dịch trung bình
          avg_active_transaction_days=("active_transaction_days", "mean"), # số ngày có giao dịch trung bình trong 1 tháng
          avg_health_score=("financial_health_score", "mean"),  # điểm sức khỏe tài chính trung bình -> xác định điểm thấp nhất
          avg_essential_ratio=("essential_spend_ratio", "mean"),  # tỷ lệ chi tiêu thiết yếu trung bình
          avg_spend_income_ratio=("spend_to_income_ratio", "mean"), # tỷ lệ chi tiêu trên thu nhập trung bìn
          customer_months=("consumer_id", "size") # quy mô mẫu tính theo số lượng khách hàng-tháng
      )
)
age_summary["pct_of_total"] = (
    age_summary["customer_months"] / age_summary["customer_months"].sum() * 100 # tỷ lệ % đóng góp của từng nhóm tuổi trong tổng thể dữ liệu
)

display(age_summary.round(3))
```

```python
# kiểm định thống kê
from scipy import stats
groups_for_anova = [
    df_balanced[df_balanced["age_cohort"] == c]["financial_health_score"].dropna()
    for c in cohort_order
]
f_stat, p_value = stats.f_oneway(*groups_for_anova)
print(f"\nANOVA financial_health_score theo age_cohort: F={f_stat:.3f}, p={p_value:.4f}")

if p_value < 0.05:
    print("Có khác biệt có ý nghĩa thống kê giữa các cohort tuổi.")
else:
    print("Không có khác biệt có ý nghĩa thống kê (p >= 0.05).")
    print("Age không phải yếu tố phân biệt mạnh cho financial health trong dataset này.")
```

```python
all_scores = df_balanced["financial_health_score"].dropna()
grand_mean = all_scores.mean()

ss_between = sum(len(g) * (g.mean() - grand_mean)**2 for g in groups_for_anova)
ss_total = ((all_scores - grand_mean)**2).sum()
eta_squared = ss_between / ss_total

print(f"F={f_stat:.3f}, p={p_value:.6f}")
print(f"Eta-squared: {eta_squared:.5f}")
print(f"=> age_cohort giải thích được {eta_squared*100:.3f}% biến thiên của health score")

if eta_squared < 0.01:
    print("Effect size RẤT NHỎ - dù p có ý nghĩa thống kê, age gần như không có giá trị thực tiễn để phân khúc khách hàng.")
elif eta_squared < 0.06:
    print("Effect size nhỏ - có tồn tại khác biệt nhưng khiêm tốn.")
else:
    print("Effect size đáng kể - age là yếu tố phân biệt thực sự quan trọng.")
```

Age không phải driver của Financial Health

Kiểm định ANOVA cho thấy `age_cohort` có khác biệt thống kê với `financial_health_score` ($F=5.509, p < 0.001$), nhưng effect size ($eta^2$ = 0.0025) cho thấy tuổi tác chỉ giải thích được $0.25\%$ sự khác biệt về sức khỏe tài chính giữa các khách hàng - mức độ ảnh hưởng không đáng kể về mặt thực tiễn.
